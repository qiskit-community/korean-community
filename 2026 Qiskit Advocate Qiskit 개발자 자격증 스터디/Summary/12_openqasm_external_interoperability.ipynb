```python
from qiskit import QuantumCircuit
from qiskit import qasm2, qasm3
```

## 1. Qiskit circuit 만들기

먼저 간단한 Bell circuit을 만든다.


```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

qc.draw("text")
```

## 2. OpenQASM 3로 export

OpenQASM 3는 dynamic circuit과 더 현대적인 classical 표현을 담기 위한 형식이다.  
Qiskit에서는 `qasm3.dumps(circuit)`으로 문자열을 만들 수 있다.


```python
qasm3_text = qasm3.dumps(qc)
print(qasm3_text)
```

## 3. OpenQASM 2 문자열을 Qiskit circuit으로 import

OpenQASM 2는 더 단순한 형식이다.  
간단한 gate와 measurement를 외부 텍스트 형식으로 주고받을 때 유용하다.


```python
qasm2_text = """
OPENQASM 2.0;
include "qelib1.inc";
qreg q[2];
creg c[2];
h q[0];
cx q[0], q[1];
measure q -> c;
"""

loaded = qasm2.loads(qasm2_text)
loaded.draw("text")
```

## 4. OpenQASM 2로 export

간단한 circuit은 OpenQASM 2로도 내보낼 수 있다.  
단, OpenQASM 2는 Qiskit object를 완전히 보존하는 일반 serialization format은 아니다.


```python
qasm2_text_from_qiskit = qasm2.dumps(qc)
print(qasm2_text_from_qiskit)
```

## Summary

| Task | Main API |
|---|---|
| Qiskit → OpenQASM 3 string | `qasm3.dumps(qc)` |
| OpenQASM 2 string → Qiskit | `qasm2.loads(text)` |
| Qiskit → OpenQASM 2 string | `qasm2.dumps(qc)` |
| file에서 읽기 | `qasm2.load(filename)` |
| file로 쓰기 | `qasm3.dump(qc, stream)` |

자격증 관점에서는 OpenQASM이 **외부 tool과 circuit을 교환하기 위한 텍스트 형식**이라는 점과, Qiskit의 `qasm2`, `qasm3` module 이름을 기억하면 충분하다.
