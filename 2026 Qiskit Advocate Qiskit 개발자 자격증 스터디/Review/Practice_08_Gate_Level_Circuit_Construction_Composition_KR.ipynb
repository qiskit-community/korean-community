- **B.** `QuantumCircuit(3, 3)`  *(첫 번째 인자 큐비트, 두 번째 인자 클래식 비트)*
- **C.** `QuantumCircuit(QuantumRegister(3))`

### 문제 2

`append`와 `compose`의 차이에 대한 설명으로 **옳은 것**은?

- **A.** `compose`는 다른 회로를 **펼쳐서(flatten)** 현재 회로의 연산 순서 안에 같은 수준의 게이트로 넣는다.
- **B.** `append`는 항상 **한 덩어리의 불투명(opaque) 연산**으로만 들어간다.
- **C.** `compose`는 서브회로 안의 게이트를 전부 단일 블록 박스로 유지한다.
- **D.** 두 메서드는 구현 이름만 다르고 동작은 항상 같다.

### 문제 3

다음 코드 실행 후 회로에서 **실제 적용 순서**(위에서부터 또는 리스트 순서대로 시간 순)로 가장 알맞은 것은?

- **A.** `CX`가 먼저, 그다음 `H`
- **B.** `H`가 먼저, 그다음 `CX`
- **C.** 순서가 정의되지 않는다.
- **D.** 회로 깊이가 0이다.

````python
from qiskit import QuantumCircuit

qc1 = QuantumCircuit(2)
qc1.h(0)

qc2 = QuantumCircuit(2)
qc2.cx(0, 1)

qc = qc1.compose(qc2, qubits=[0, 1])
print(qc)
qc.draw("mpl")
````

### 문제 7

다음 중 **에러 없이 실행되는 코드**만 **모두** 고르시오 (`from qiskit import QuantumCircuit`, `from qiskit.circuit.library import XGate` 가 있다고 가정).

- **A.** `QuantumCircuit(1).append(XGate(), [0], [])`
- **B.** `QuantumCircuit(1).append(XGate(), 0)`
- **C.** `QuantumCircuit(1).compose(XGate(), [0])`
- **D.** `QuantumCircuit(1).append(XGate(), qargs=0)`

**(참고)** Qiskit 쪽에서는 `compose`의 첫 인자로 회로만이 아니라 `Gate`(및 다른 `Operation`)도 받도록 확장되어 있어, 단일 게이트를 회로 레이아웃 위에 놓아 합치는 패턴이 유효하다.

### 문제 8

역회로를 만들 때 다음 코드의 결과에 가깝게 **옳은** 설명은?

- **A.** `qc_inv`는 큐비트 레이아웃을 유지한 채 **게이트 순서를 거꾸로** 놓고 **각 게이트의 역**을 취한 순수 유니터리 회로이다(합성 순서까지 고려하면 `qc`의 수학적 역과 대응).
- **B.** `inverse()`는 클래식 측정이 있어도 항상 유효하고 같은 비트 레이아웃을 유지한다.
- **C.** `inverse()` 후에도 회로 깊이(`depth`)는 항상 1이다.
- **D.** `H` 게이트는 역행렬이 없으므로 오류가 난다.

````python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc_inv = qc.inverse()
print(qc_inv)
qc_inv.draw("mpl")
````

### 문제 9

두 회로를 **서로 다른 큐비트 부분공간에** 놓여 있게 매핑하며 하나로 합치고 싶다. 다음 설명 중 **실제 사용 패턴과 맞는** 것은?

- **A.** `combined`는 4큐비트이며 `left`는 큐비트 0–1, `right`는 2–3에 해당한다.
- **B.** `compose` 두 번 호출하면 두 번째 호출이 첫 번째를 덮어쓴다.
- **C.** `qubits=[0,1]`은 “전치 행렬”을 의미하므로 순서만 바뀐다.
- **D.** 같은 길이의 서브회로는 **한 번에 하나의 연속 구간에만** 붙일 수 있다.

````python
from qiskit import QuantumCircuit

left = QuantumCircuit(2)
left.h(0)

right = QuantumCircuit(2)
right.x(1)

combined = QuantumCircuit(4).compose(left, qubits=[0, 1]).compose(right, qubits=[2, 3])
print(combined)
combined.draw("mpl")
````

### 문제 10

부분 회로를 **한 블록(박스)으로 유지한 채** `append` 하는 일반 패턴과 가장 가까운 코드는?

- **A.**
  ```python
  sub = QuantumCircuit(2)
  sub.cx(0, 1)

  qc = QuantumCircuit(2)
  qc.append(sub.to_gate(), [0, 1])
  ```
- **B.**
  ```python
  sub = QuantumCircuit(2)
  sub.cx(0, 1)

  qc = QuantumCircuit(2)
  qc.compose(sub, qubits=[0, 1])
  ```
- **C.**
  ```python
  qc = QuantumCircuit(2)
  qc.append(sub, [0, 1])  # 정의 없이 호출만
  ```

- **D.**
  ```python
  sub = QuantumCircuit(2)
  qc = QuantumCircuit(2)
  qc.inverse(sub.to_gate())
  ```

| 6 | **B** | 유니터리 블록을 게이트로 쓰려면 보통 `to_gate()`가 맞고, 측정 등 비유니터리 포함 시 `to_instruction()` 등 패턴으로 나뉜다는 점을 시험에서 자주 접한다. |

| 10 | **A** | 서브회로를 **단일 블록**으로 보이려면 `to_gate()`(또는 `to_instruction()`) 후 `append`가 전형적이다. B의 `compose`는 서브회로 게이트를 **펼쳐 넣는** 쪽이다. |
