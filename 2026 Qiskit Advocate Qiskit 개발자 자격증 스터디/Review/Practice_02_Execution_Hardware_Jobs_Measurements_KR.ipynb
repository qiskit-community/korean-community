## 정답 및 해설 요약

| 번호 | 정답 | 요지 |
|:---:|:---:|------|
| 1 | **B** | 회로 = 무엇을 계산할지; Job/run 파이프라인 = 언제·어디서·어떻게 제출·추적·회수할지. |
| 2 | **C** | `JobStatus`: `INITIALIZING`, `QUEUED`, `VALIDATING`, `RUNNING`, `CANCELLED`, `DONE`, `ERROR` 등 — **`COMPLETED`는 없음**(정상 종료는 `DONE`). |
| 3 | **B** | 트랜스파일 목적 = ISA(게이트 집합·토폴로지 등) 맞추기 — `S4-001`과 동일 정신. |
| 4 | **A** | 물리 레이아웃이 바뀌면 Pauli observable도 `apply_layout`으로 같이 옮겨야 함. |
| 5 | **B** | Estimator PUB는 (회로, observable들 [, 매개변수]). |
| 6 | **D** | Sampler V2: `result[i].data.meas.get_counts()` → `BitArray` 기준 비트스트링별 카운트 dict. |
| 7 | **B** | Batch: 독립 Job 묶음, 순서 보장 없음, 캘리브·타 사용자 Job이 끼일 수 있음. |
| 8 | **B** | 브로드캐스팅: **회로 템플릿은 1개**(한 PUB에 하나의 `QuantumCircuit`), 내부에서 3×100 기대값 펼침 → `evs.shape` **`(3, 100)`**. 100·300개 회로를 사용자가 따로 만들지 않는다. |
| 9 | **C** | `SWAP`→`CX` 등 **분해·번역**은 **트랜스파일(transpile)** 파이프라인(예: basis translator) 안에서 이뤄진다. **routing**은 연결도 때문에 SWAP **삽입**에 가깝고(S4-002), **scheduling**은 시간 배치, **optimization**은 별도 목적. |
| 10 | **B** | Sampler ≈ 표본/측정 통계, Estimator ≈ 기대값 추정. |
